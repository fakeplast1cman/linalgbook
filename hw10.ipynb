```python
#11.1-11.2
import matplotlib.pyplot as plt
A = np.float32(
        np.array(
             [[4, 25],
              [12, 54],
              [3, 21],
              [13, 75],
              [7, 48],
              [12, 54],
              [3, 21],
              [5, 29],
              [12, 54],
              [6, 35],
              [18, 120]]))

Y = A[:, 1]
X = np.hstack((np.ones((len(Y), 1)), A[:, 0].reshape(len(Y), 1)))

print(X)
Q, R = np.linalg.qr(X)

b = np.linalg.inv(R) @ Q.T @ Y

err = (Y - X @ b) 

dot_value = err @ X

plt.plot(err, X @ b, 'ko')
dot_value, np.corrcoef(err, X[:, 1])
```

```python
def result(A):
    X = np.hstack((np.ones((A.shape[0], 1)), A[:,0].reshape(A.shape[0], 1)))
    Y = A[:, 1]
    Q, R = np.linalg.qr(X)
    B = np.linalg.inv(R) @ Q.T @ Y
    return [X @ B, Y]

A1 = np.float32(
        np.array(
             [[4, 25],
              [12, 54],
              [3, 21],
              [13, 75],
              [7, 48],
              [12, 54],
              [3, 21],
              [5, 29],
              [12, 54],
              [6, 35],
              [18, 120]]))

A2 = np.float32(
        np.array(
             [[4, 25],
              [12, 54],
              [3, 121],
              [13, 75],
              [7, 48],
              [12, 54],
              [3, 21],
              [5, 129],
              [12, 54],
              [6, 35],
              [18, 120]]))

A3 = np.float32(
        np.array(
             [[4, 25],
              [12, 54],
              [3, 21],
              [13, 75],
              [7, 148],
              [12, 54],
              [3, 21],
              [5, 129],
              [12, 54],
              [6, 35],
              [18, 120]]))

pred1 = result(A1)
pred2 = result(A2)
pred3 = result(A3)


X = [A1[:,0], A2[:,0], A3[:,0]]
XB = [pred1, pred2, pred3]
Y = [A1[:,1], A2[:,1], A3[:,1]]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, ax in enumerate(axes):
    
    ax.plot(X[idx], XB[idx][0])


for idx, ax in enumerate(axes):
    
    ax.scatter(X[idx], Y[idx], zorder=1)
    
```
