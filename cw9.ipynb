```python
import numpy as np
import sympy as sym
```

```python
A = np.random.randn(4, 4)
B = np.random.randn(4, 4)
A @ (np.linalg.inv(A) @ B) - B
```

```python
# Нахождение обратной матрицы через rref 

A = np.random.randn(2, 4) 
A[:,2:4] = np.eye(2, 2)
A_inv = np.linalg.inv(A[:, :2])
sym_a = sym.Matrix(A).rref()
sym_a[0], A_inv
```

```python
import scipy.linalg as sl

A = np.random.randn(4, 4)
_, L, U = sl.lu(A)
_, L, U, A @ np.linalg.inv(L) 
```
