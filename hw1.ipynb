```python
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
vec1 = np.array([[7, 3], [3, 5], np.array([7, 3]) - np.array([3, 5])])

xs = vec1[:, 0]
ys = vec1[:, 1]

x_start = np.zeros(len(xs))
y_start = np.zeros(len(ys))

fig, ax = plt.subplots(figsize=(5, 5))

plt.xlim(-1, 10)
plt.ylim(-10, 10)
plt.grid(True, linestyle='--')
plt.axvline(0)
plt.axhline(0)
ax.quiver(x_start, y_start, xs, ys, angles='xy', scale_units='xy', scale=1)

plt.show()
```

```python
vec = np.array([4, 9])
norm = math.sqrt(sum([i**2 for i in vec]))

norm - np.linalg.norm(vec)
```

```python
def edv(vec):
    return 1 / np.linalg.norm(vec) * vec

edv(np.array([5, 8, 7]))
```

```python
def some_v(vec, module):
    return module / np.linalg.norm(vec) * vec

some_v(np.array([5, 8, 7]), 2)
```

```python
def transpose(vec):
    tr_vec = np.zeros(shape=(len(vec), 1))
    for idx, elem in enumerate(vec):
        tr_vec[idx, 0] = elem
    return tr_vec

transpose(np.array([4, 5, 7]))
```

```python
value = np.array([5, 8, 7])

np.linalg.norm(value)**2 - np.dot(value, value)
```

```python
value1 = np.array([5, 8, 7])
value2 = np.array([2, 4, 1])

value1*value2.T - value2*value1.T
```

```python
vec1 = np.array([7, 7])
vec2 = np.array([6, 1])

b = np.dot(vec1, vec2.T) / np.dot(vec1, vec1.T)
betta_vecs = vec1*b
vecs = np.array([vec1, vec2, betta_vecs])
perp = vec2 - betta_vecs
vec_x = vecs[:, 0]
vec_y = vecs[:, 1]
start_x = np.zeros(len(vecs))
start_y = np.zeros(len(vecs))

fig, ax = plt.subplots(figsize=(5, 5))
plt.grid(True)
plt.xlim(-5, 10)
plt.ylim(-5, 10)
plt.axhline(0)
plt.axvline(0)
ax.quiver(start_x, start_y, vec_x, vec_y, angles='xy', scale_units='xy', scale=1)
ax.quiver(betta_vecs[0], betta_vecs[1], perp[0], perp[1], angles='xy', scale_units='xy', scale=1, color='blue')

plt.show()
```

```python
vec1 = np.array([6, 3])
vec2 = np.array([8,-3])

sub_vec2 = np.dot(vec2, vec1.T) / np.dot(vec2, vec2.T) * vec2
vecs = np.array([vec1, vec2, sub_vec2, vec1-sub_vec2])

fig, ax = plt.subplots(figsize=(6, 6))

plt.grid(True)
plt.xlim(-6, 10)
plt.ylim(-5, 10)
ax.quiver(np.zeros(len(vecs)), np.zeros(len(vecs)), vecs[:, 0], vecs[:, 1], scale=1, angles='xy', scale_units='xy')

plt.show()
```
