Кластеризация методом k-средних.

```python
data = np.array([np.random.uniform(-100, 110, 150), np.random.uniform(-90, 150, 150)]).T

k = 3
ridx = np.random.choice(range(len(data)), k, replace=False)
centroids = data[ridx, :]
dists = np.zeros((data.shape[0], k))
print(centroids)
for i in range(50):
    for kl in range(k):
        dists[:, kl] = np.sum((data-centroids[kl, :])**2, axis=1)

    groupidx = np.argmin(dists, axis=1)
    for ki in range(k):
        centroids[ki, :] = [np.mean(data[groupidx==ki, 0]), np.mean(data[groupidx==ki, 1])]

    print(centroids)
```
