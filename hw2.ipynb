```python
some_points1 = np.array([1, 3])
xscalars = np.random.uniform(-4, 4, 100)


x = np.array(list(map(lambda v: some_points1 * v, xscalars)))
fig = go.Figure(data=[[go.Scatter3d(
                        x = x[:, 0],
                        y = x[:, 1]
                    )], 
]
)
fig.show()
```

```python
some_points2 = np.array([1, 0, 0])
some_points3 = np.array([0, 1, 0])
some_points4 = np.array([0, 0, 1])
yscalars = np.random.uniform(-4, 4, 1000)
zscalars = np.random.uniform(-4, 4, 1000)
ascalars = np.random.uniform(-4, 4, 1000)
y = np.array(list(map(lambda v: some_points2 * v, yscalars)))
z = np.array(list(map(lambda v: some_points3 * v, zscalars)))
a = np.array(list(map(lambda v: some_points4 * v, ascalars)))
z = z + y + a

fig = go.Figure(data=[
    go.Scatter3d(x=z[:, 0], y=z[:, 1], z=z[:, 2], mode='markers')
])

fig.show()
```
